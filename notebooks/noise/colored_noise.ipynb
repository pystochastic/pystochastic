```python
import numpy as np
import plotly.graph_objs as go
```

```python
from pystochastic.processes.noise import ColoredNoise as TestType
from stochastic.noise import ColoredNoise as StochasticTestType
```

```python
seed = 1234567890
n = 10
t = 1
beta = 0

rng = np.random.default_rng(seed)
np.random.seed(seed)
```

```python
x = TestType(beta=beta, t=t, rng=rng)
xs = x.sample(n=n)

y = StochasticTestType(beta=beta, t=t)
ys = y.sample(n=n)
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        y=xs,
        mode="lines",
        name="Current",
    )
)

fig.add_trace(
    go.Scatter(
        y=ys,
        mode="lines",
        name="Original",
    )
)


fig.update_layout(
    title="Samples",
    xaxis_title="Sample Index",
    yaxis_title="Amplitude",
)

fig.show()
```
