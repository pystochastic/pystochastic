```python
import numpy as np
import plotly.graph_objs as go
```

```python
from pystochastic.processes.noise import FractionalGaussianNoise as TestType
from stochastic.noise import FractionalGaussianNoise as StochasticTestType
```

```python
seed = 1234567890
n = 10
t = 1
hurst = 0.15

rng = np.random.default_rng(seed)
np.random.seed(seed)
```

```python
x = TestType(hurst=hurst, t=t, rng=rng)
xs = x.sample_davies_harte(n=n)


y = StochasticTestType(hurst=hurst, t=t)
ys = y.sample(n=n, algorithm="daviesharte")
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        y=xs,
        mode="lines",
        name="Current",
    )
)

fig.add_trace(
    go.Scatter(
        y=ys,
        mode="lines",
        name="Original",
    )
)


fig.update_layout(
    title="Samples",
    xaxis_title="Sample Index",
    yaxis_title="Amplitude",
)

fig.show()
```
