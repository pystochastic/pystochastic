```python
import numpy as np
import plotly.graph_objs as go
```

```python
from pystochastic.processes.continuous import (
    MultifractionalBrownianMotion as TestType,
)
from stochastic.continuous import MultifractionalBrownianMotion as StochasticTestType
```

```python
seed = 1234567890
n = 10
t = 1
hurst = lambda t: 0.1 + 0.8 * t / t

rng = np.random.default_rng(seed)
np.random.seed(seed)
```

```python
x = TestType(hurst=hurst, t=t, rng=rng)
xs = x.sample(n=n)


y = StochasticTestType(hurst=hurst, t=t)
ys = y.sample(n=n)
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        y=xs,
        mode="lines",
        name="Current",
    )
)

fig.add_trace(
    go.Scatter(
        y=ys,
        mode="lines",
        name="Original",
    )
)


fig.update_layout(
    title="Samples",
    xaxis_title="Sample Index",
    yaxis_title="Amplitude",
)

fig.show()
```
