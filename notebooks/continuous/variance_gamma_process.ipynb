```python
import numpy as np
import plotly.graph_objs as go
```

```python
from pystochastic.processes.continuous import VarianceGammaProcess as TestType
from stochastic.continuous import VarianceGammaProcess as StochasticTestType
```

```python
seed = 1234567890
n = 10
t = 1
drift = 0.2
variance = 0.5
scale = 1.3

rng = np.random.default_rng(seed)
np.random.seed(seed)
```

```python
x = TestType(drift=drift, variance=variance, scale=scale, t=t, rng=rng)
xs = x.sample(n=n)

y = StochasticTestType(drift=drift, variance=variance, scale=scale, t=t)
ys = y.sample(n=n)
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        y=xs,
        mode="lines",
        name="Current",
    )
)

fig.add_trace(
    go.Scatter(
        y=ys,
        mode="lines",
        name="Original",
    )
)


fig.update_layout(
    title="Samples",
    xaxis_title="Sample Index",
    yaxis_title="Amplitude",
)

fig.show()
```

```python
display(xs, ys)
```
