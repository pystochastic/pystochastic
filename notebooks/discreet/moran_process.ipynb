```python
import numpy as np
import plotly.graph_objs as go
```

```python
from pystochastic.processes.discrete import MoranProcess as TestType
from stochastic.discrete import MoranProcess as StochasticTestType
```

```python
seed = 1234567890
n = 5
maximum = 10
start = 5

rng = np.random.default_rng(seed)
np.random.seed(seed)
```

```python
x = TestType(
    maximum=maximum,
    rng=rng,
)
xs = x.sample_with_start(n=n, start=start)

y = StochasticTestType(
    maximum=maximum,
)
ys = y.sample(n=n, start=start)
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        y=xs,
        mode="lines",
        name="Current",
    )
)

fig.add_trace(
    go.Scatter(
        y=ys,
        mode="lines",
        name="Original",
    )
)


fig.update_layout(
    title="Samples",
    xaxis_title="Sample Index",
    yaxis_title="Amplitude",
)

fig.show()
```
